```python
import pandas as pd
import numpy as np
df = pd.read_csv("Telco_Customer_Churn_Dataset .csv")
print("Dataset loaded successfully!")
```

```python
df.head()
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
df.info()
```

```python
missing_values = df.isnull().sum()
missing_values = missing_values[
    missing_values > 0
].sort_values(ascending=False)
print("Columns containing missing values:")
print(missing_values)
```

```python
blank_totalcharges = (
    df['TotalCharges']
    .astype(str)
    .str.strip()
    .eq('')
    .sum())
print( "Blank values in TotalCharges:",
    blank_totalcharges)
```

```python
df['TotalCharges'] = pd.to_numeric(
    df['TotalCharges'],
    errors='coerce')
print(df['TotalCharges'].dtype)
```

```python
print(  "Missing TotalCharges after conversion:",
    df['TotalCharges'].isnull().sum())
df['TotalCharges'] = df['TotalCharges'].fillna(0)
```

```python
print( "Missing TotalCharges after treatment:",
    df['TotalCharges'].isnull().sum())
```

```python
duplicate_count = df.duplicated().sum()
print( "Number of duplicate rows:",
    duplicate_count)
```

```python
df.dtypes
```

```python
categorical_columns = df.select_dtypes(
    include=['object']
).columns.tolist()
print("Categorical columns:")
print(categorical_columns)
```

```python
numerical_columns = df.select_dtypes(
    include=np.number
).columns.tolist()
print("Numerical columns:")
print(numerical_columns)
```

```python
for column in categorical_columns:
    print("\n", column)
    print(df[column].unique())
```

```python
model_data = df.drop(
    columns=['customerID']
).copy()
print("Shape after removing customerID:")
print(model_data.shape)
```

```python
df
```

```python
encoded_data = pd.get_dummies(
    model_data,
    columns=[
        col for col in model_data.select_dtypes(
            include=['object']
        ).columns ],
    drop_first=True,
    dtype=int)
```

```python
print("Original cleaned dataset shape:")
print(df.shape)
print("\nEncoded dataset shape:")
print(encoded_data.shape)
```

```python
encoded_data.head()
```

```python
print("FINAL DATA QUALITY CHECK ")
print("Rows:", encoded_data.shape[0])
print("Columns:", encoded_data.shape[1])
print( "Missing values:",
    encoded_data.isnull().sum().sum())
print(  "Duplicate rows:",
    encoded_data.duplicated().sum())
print( "Non-numeric columns:",
    encoded_data.select_dtypes(
        include=['object']
    ).columns.tolist())
```

```python
cleaning_summary = pd.DataFrame({
    'Check': [
        'Initial rows',
        'Initial columns',
        'Blank TotalCharges',
        'Missing values after cleaning',
        'Duplicate rows',
        'customerID removed from model data',
        'Categorical variables encoded'],
    'Result': [
        df.shape[0],
        df.shape[1],
        blank_totalcharges,
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        'Yes',
        'Yes' ]})
cleaning_summary
```
